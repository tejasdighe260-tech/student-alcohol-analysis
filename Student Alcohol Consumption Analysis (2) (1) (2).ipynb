## Alcohol Consumption & Social Factors

```python
# distrubution of alcohol comsumption by Gender

df.groupby('sex')[['Dalc', 'Walc']].mean().plot(kind='bar')

plt.title('Average Alcohol Consumption by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Alcohol Consumption')
plt.xticks(rotation=0)
plt.legend(['Weekday', 'Weekend'])
plt.show()
```

```python
age_gender = df.groupby(['age', 'sex'])[['Dalc', 'Walc']].mean().reset_index()
```

```python
# Which Age Group has highest alcohol consumption by gender

age_gender = df.groupby(['age', 'sex'])['Walc'].mean().reset_index()

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=age_gender,
    x='age',
    y='Walc',
    hue='sex',
    marker='o'
)

plt.title('Age-wise Weekend Alcohol Consumption by Gender')
plt.xlabel('Age')
plt.ylabel('Average Weekend Alcohol Consumption')
plt.legend(title='Gender')
plt.grid(True)
plt.show()
```

### Inferences

1. Male students generally have higher average weekend alcohol consumption than female students across most age groups.
2. For females, consumption remains fairly stable between ages 15–20, at around 1.8–2.2 units, but decreases to about 1.0 unit at age 21.
3. For males, consumption increases from age 15 to 17, reaching about 3.2 units, and remains almost the same at age 18.

### Alcohol Consumption & Student Outcomes

```python
# how does weekend alcohol consumption relate to Students final grade

sns.lmplot(
    data=df,
    x='absences',
    y='G3',
    height=5,
    aspect=1.6,
    fit_reg=True,
    hue='Walc',
    ci=95
)

plt.suptitle('Absences vs Final Grades by Weekend Alcohol Consumption',
             y=1.02)

plt.xlabel('Number of Absences')
plt.ylabel('Final Grade (G3)')
plt.show()
```

### Inferences

1. There is a generally negative relationship between absences and final grade (G3) for several alcohol-consumption groups.
       As the number of absences increases, final grades tend to decrease slightly.
2. The relationship is not very strong. The points are widely scattered, 
       meaning students with similar numbers of absences can have quite different final grades.

3. For Walc = 1, the regression line shows a noticeable downward trend, suggesting that more absences are associated with lower final grades.
4. For Walc = 2 and Walc = 4, the trend is also generally downward, although the relationship appears relatively weak.
5. For Walc = 3, the regression line is slightly upward. This indicates that the relationship between absences and final grade is not consistent across all alcohol-consumption levels.
6. Walc = 5 has relatively limited visible data at higher absence levels, so its trend should be interpreted cautiously.

```python
# how does final grade differ between math and Portuguese

sns.lmplot(
    data=df,
    x='Walc',
    y='G3',
    col='course',
    height=5,
    aspect=1.2,
    ci=95,
    scatter_kws={'alpha': 0.5}
)

plt.show()
```

### Inferences


1. There is a weak negative relationship between weekend alcohol consumption (Walc) and final grade (G3) in both courses.  
    As Walc increases, the regression line generally slopes downward.
2. Math: The trend is only slightly negative. Average G3 appears to decrease gradually as weekend alcohol consumption   
      increases from level 1 to level 5.
3. Portuguese: The negative trend is somewhat more noticeable than in Math. Students with higher Walc levels tend to have slightly lower final grades.
